```python
import cv2
import os
```

# Extract video

```python


#  extract frames from a video
def extract_frames(video_file):
    cap = cv2.VideoCapture(video_file)
    
    frame_rate = 8  # frame rate
    frame_count = 0
    
    video_name = os.path.splitext(os.path.basename(video_file))[0]

    output_directory = f"PoC_frames_{video_name}"
    os.makedirs(output_directory, exist_ok=True)
        
    while True:  # we manually extract at the desired frame rate
        ret, frame = cap.read()
        
        if not ret:
            break
        
        frame_count += 1
        
        if frame_count % int(cap.get(5) / frame_rate) == 0:
            output_file = f"{output_directory}/frame_{frame_count}.jpg"
            cv2.imwrite(output_file, frame)
            print(f"Frame {frame_count} has been extracted and saved as {output_file}")
    
    cap.release()


```
